## 2. Loading the Dataset

In this section, we load the raw datasets into Python using the Pandas library. The project consists of four datasets: Customers, Orders, Order Items, and Products.

These datasets contain information about customer details, order transactions, purchased products, and product pricing. We will load each dataset separately and perform an initial inspection to understand its structure and identify potential data quality issues.

```python
orders = pd.read_csv("../data/raw/orders.csv", parse_dates=["order_date"])
order_items = pd.read_csv("../data/raw/order_items.csv")
products = pd.read_csv("../data/raw/products.csv")
customers = pd.read_csv("../data/raw/customers.csv", parse_dates=["signup_date"])
```

## 3. Understanding the Dataset Structure

In this section, we will examine the basic structure of all four datasets to understand their dimensions, columns, data types, and sample records.

The objective is to understand how the data is organized before performing data quality checks and cleaning.

### 3.1 Dataset Dimensions

In this step, we examine the dimensions of each dataset to understand the total number of records and columns available for analysis. This provides an overview of the dataset size and structure.

```python
data = {"orders":orders,"oeder_items":order_items,"products":products,"customers":customers}
for name, df in data.items():
    print(f"{name}:{df.shape}")

```

### 3.5 Duplicate Records Inspection

In this step, we check each dataset for completely duplicated rows. Identifying duplicates is important to prevent potential data inaccuracies and double counting during analysis. Any detected duplicates will be investigated before removal.

```python
for name, df in data.items():
    print(f"{name}\n:{df.duplicated().sum()}")
```

### 3.6 Missing Values Inspection

In this step, we identify the number of missing values in each column across all four datasets. Missing values will be investigated to understand their possible causes and determine the appropriate treatment without making assumptions.

```python
for name, df in data.items():
    print(f"\n{name}\n:{df.isnull().sum()}")
```
