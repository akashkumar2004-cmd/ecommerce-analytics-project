### Historical Price Comparison – Findings

I compared historical `unit_price` values with the current `list_price` for products associated with missing prices.

**Findings:**
- The same product can have different historical selling prices.
- Historical prices do not always match the current catalog price.
- Therefore, using `list_price` to fill missing historical prices could introduce inaccurate revenue calculations.

**Decision:** I will not replace missing `unit_price` values with current catalog prices. Further investigation is required before deciding how to handle these records.

```python
# Merge all order items with product information
price_comparison = order_items.merge(
    products[["product_id", "product_name", "list_price"]],
    on="product_id",
    how="left"
)

# Calculate difference between historical unit price and list price
price_comparison["price_difference"] = (
    price_comparison["unit_price"] -
    price_comparison["list_price"]
)

# Inspect available prices for products with missing prices
missing_products = (
    price_comparison[
        price_comparison["unit_price"].isnull()
    ]["product_id"]
    .unique()
)

for product_id in missing_products[:5]:
    print(f"\nProduct ID: {product_id}")

    print(
        price_comparison[
            (price_comparison["product_id"] == product_id) &
            (price_comparison["unit_price"].notnull())
        ][["unit_price", "list_price", "price_difference"]]
        .drop_duplicates()
        .head(10)
    )
```

### Historical Price Variation – Findings

I compared historical selling prices with current catalog prices.

**Observations:**
- Multiple historical selling prices exist for the same product.
- In the inspected examples, historical prices are approximately 5% higher than current list prices.
- This indicates that current catalog prices cannot automatically be treated as historical transaction prices.

**Decision:** Further investigation by product and year is required before handling missing unit prices.

```python
# Check historical price variation by product and year
price_patterns = (
    price_comparison[
        price_comparison["unit_price"].notnull()
    ]
    .groupby(["product_id", "list_price"])
    ["unit_price"]
    .agg(["min", "max", "nunique"])
    .reset_index()
)

print(price_patterns.head(20))
```

### Historical Pricing – Yearly Analysis

I grouped available historical unit prices by product and year.

**Initial observations:**
- The inspected products have one unique selling price in 2024.
- The same products have a different selling price in 2025.
- In the inspected examples, 2025 prices are approximately 5% higher than 2024 prices.
- The 2024 prices match the current catalog prices in the inspected examples.

**Next step:** Verify whether this pricing pattern applies consistently across all products before considering historical price recovery.

```python
# Add order date and year to the price comparison
price_comparison = order_items.merge(
    orders[["order_id", "order_date"]],
    on="order_id",
    how="left"
)

price_comparison = price_comparison.merge(
    products[["product_id", "list_price"]],
    on="product_id",
    how="left"
)

price_comparison["year"] = (
    price_comparison["order_date"].dt.year
)

# Analyze historical prices by product and year
price_patterns_by_year = (
    price_comparison[
        price_comparison["unit_price"].notnull()
    ]
    .groupby(["product_id", "year"])
    ["unit_price"]
    .agg(["min", "max", "nunique"])
    .reset_index()
)

print(price_patterns_by_year.head(20))
```

### Historical Pricing – Final Investigation Results

I verified historical unit prices across all 90 products for 2024 and 2025.

**Findings:**
- All 90 products have observed prices in both years.
- Each product has exactly one unique observed selling price within each year.
- No products have multiple selling prices within the same year.
- The inspected examples show a consistent 5% price increase in 2025 compared with 2024.

**Cleaning decision:**

The 199 missing `unit_price` values can be reconstructed using the observed historical price for the matching `product_id` and year.

I will preserve the original missing-value information by creating a separate indicator column to identify reconstructed prices.

```python
# Check whether every product has exactly one price per year

price_variation_check = (
    price_patterns_by_year
    .groupby("year")["nunique"]
    .value_counts()
    .sort_index()
)

print(price_variation_check)

# Check products with multiple prices in the same year
inconsistent_prices = price_patterns_by_year[
    price_patterns_by_year["nunique"] > 1
]

print("\nProducts with multiple prices in the same year:")
print(inconsistent_prices)
```

```python
year_coverage = (
    price_patterns_by_year
    .groupby("product_id")["year"]
    .nunique()
)

print("Products with both years:", (year_coverage == 2).sum())
print("Products without both years:", (year_coverage < 2).sum())
```

### Missing Unit Price – Treatment

After investigating historical prices, I created a lookup table containing 180 product-year price combinations.

All 90 products had exactly one observed historical selling price in both 2024 and 2025.

I used this lookup table to reconstruct the 199 missing `unit_price` values by matching `product_id` and order year.

A `price_was_missing` indicator column was created to preserve information about the original missing values.

The reconstructed prices will be validated before being used in revenue calculations.

```python
# Create historical price lookup
historical_prices = (
    price_comparison[
        price_comparison["unit_price"].notnull()
    ]
    .groupby(["product_id", "year"])["unit_price"]
    .first()
    .reset_index()
)

print(historical_prices.head(10))
print("Historical price records:", len(historical_prices))
```

### Missing Unit Price – Cleaning and Validation Results

I reconstructed the 199 missing `unit_price` values using historical prices matched by product ID and order year.

**Results:**
- Original missing prices: 199
- Reconstructed prices: 199
- Remaining missing prices: 0
- Total order-item records were retained.

I preserved the original missing-value information using the `price_was_missing` indicator column.

The reconstructed dataset will undergo additional validation before being saved as the processed dataset.

```python
#previewing the missing values
order_items["price_was_missing"] = (
    order_items["unit_price"].isnull()
)

print(order_items["price_was_missing"].value_counts())
```

```python
#add the order year

order_items_with_year = order_items.merge(
    orders[["order_id", "order_date"]],
    on="order_id",
    how="left"
)

order_items_with_year["year"] = (
    order_items_with_year["order_date"].dt.year
)
```

```python
#merge historical price
order_items_with_year = order_items_with_year.merge(
    historical_prices,
    on=["product_id", "year"],
    how="left",
    suffixes=("", "_historical")
)
```

```python
#fill only missing price 
order_items_with_year["unit_price"] = (
    order_items_with_year["unit_price"].fillna(
        order_items_with_year["unit_price_historical"]
    )
)
```

```python
# validate the result

print(
    "Missing prices after cleaning:",
    order_items_with_year["unit_price"].isnull().sum()
)

print(
    "Reconstructed prices:",
    order_items_with_year["price_was_missing"].sum()
)
```

### Discount and Net Revenue Validation

Gross amount was calculated as:

`quantity × unit_price`

Net line-item amount was calculated as:

`gross_amount - discount_amount`

**Validation findings:**
- No discounts exceeded their corresponding gross amounts.
- No negative net amounts were identified.
- All discount values were non-negative.

**Decision:**

All records passed the discount validation. No corrections were necessary.

```python
# Calculate gross amount for each order item
order_items_with_year["gross_amount"] = (
    order_items_with_year["quantity"] *
    order_items_with_year["unit_price"]
)

# Calculate net amount after discount
order_items_with_year["net_amount"] = (
    order_items_with_year["gross_amount"] -
    order_items_with_year["discount_amount"]
)

# Check discounts exceeding gross amount
print(
    "Discounts exceeding gross amount:",
    (
        order_items_with_year["discount_amount"] >
        order_items_with_year["gross_amount"]
    ).sum()
)

# Check negative net amounts
print(
    "Negative net amounts:",
    (order_items_with_year["net_amount"] < 0).sum()
)
```

**4. Business Decision**

```python
# Attach order status to each order-item row
order_items_status = order_items_with_year.merge(
    orders[["order_id", "status"]],
    on="order_id",
    how="left",
    validate="many_to_one"
)

# Summarise merchandise value by status
status_revenue_summary = (
    order_items_status.groupby("status")
    .agg(
        order_count=("order_id", "nunique"),
        item_rows=("order_id", "size"),
        merchandise_value=("net_amount", "sum")
    )
    .reset_index()
)

print(status_revenue_summary)
```

```python
# inspecting shipping fee seperetly
shipping_summary = (
    orders.groupby("status")
    .agg(
        order_count=("order_id", "count"),
        total_shipping_fees=("shipping_fee", "sum")
    )
    .reset_index()
)

print(shipping_summary)
```
