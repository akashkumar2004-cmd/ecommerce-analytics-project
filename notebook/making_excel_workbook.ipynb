```python
import pandas as pd
from pathlib import Path

# Folder containing your CSV files
data_folder = Path("../data/raw")

# Output Excel workbook
output_file = Path("../data/processed/ecommerce_data.xlsx")

# CSV files and their worksheet names
csv_files = {
    "customers": "customers.csv",
    "orders": "orders.csv",
    "order_items": "order_items.csv",
    "products": "products.csv"
}

# Create one Excel workbook with four worksheets
with pd.ExcelWriter(output_file, engine="openpyxl") as writer:
    for sheet_name, filename in csv_files.items():
        file_path = data_folder / filename

        df = pd.read_csv(file_path)
        df.to_excel(
            writer,
            sheet_name=sheet_name,
            index=False
        )

        print(f"{filename}: {len(df)} rows exported")

print(f"\nWorkbook created: {output_file}")
```
