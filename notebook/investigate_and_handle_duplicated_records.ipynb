### Duplicate Records

During the initial data quality assessment, 38 duplicate records were identified in the Orders dataset.

After investigating the duplicate records, they were confirmed to be genuine duplicates rather than legitimate separate transactions. Therefore, these records will be removed to maintain data accuracy and prevent potential overcounting in subsequent sales analysis.

```python
orders.duplicated(subset="order_id").sum()
```

```python
#removing duplicates
orders.drop_duplicates(inplace=True)
```

```python
#verifying
orders.duplicated().sum()
```
